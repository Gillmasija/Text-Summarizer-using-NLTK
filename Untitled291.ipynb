```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
# Download NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# Ask the user for text
text = input("Enter your text: ")
# Split text into sentences
sentences = sent_tokenize(text)
# Get stopwords
stop_words = set(stopwords.words("english"))
# Count important words
word_frequency = {}
for sentence in sentences:
    words = word_tokenize(sentence.lower())
    for word in words:
        if word not in stop_words:
            if word in word_frequency:
                word_frequency[word] += 1
            else:
                word_frequency[word] = 1
```

```python
sentence_scores = {}
# Go through each sentence again
for sentence in sentences:
    # Convert the sentence to lowercase
    # and split it into words
    words = word_tokenize(sentence.lower())
    # Start the sentence score at 0
    score = 0
    # Check every word in the sentence
    for word in words:
        # If the word is in our word frequency dictionary
        if word in word_frequency:
            # Add the word's frequency to the sentence score
            # Words that appear more often give the sentence
            # a higher score
            score += word_frequency[word]
    # Save the sentence and its score
    sentence_scores[sentence] = score
```

```python
number = int(input("How many sentences do you want in the summary? "))
```

```python
summary = sorted(
    sentence_scores,
    key=sentence_scores.get,
    reverse=True
)[:number]
print("\n Summary:")
print("".join(summary))
```
